```python
import os
import kagglehub
import pandas as pd
```

```python
path = kagglehub.dataset_download("rishikeshkonapure/zomato")
file_path = os.path.join(path, "zomato.csv")
df = pd.read_csv(file_path)
```

```python
df
```

```python
df.columns
```

```python
columns_to_keep = ['name', 'online_order', 'book_table', 'rate', 'dish_liked', 'approx_cost(for two people)']
columns_to_drop = ['url', 'address', 'votes',
       'phone', 'location', 'rest_type', 'cuisines', 'reviews_list', 'menu_item',
       'listed_in(type)', 'listed_in(city)']
```

```python
df.drop(columns=columns_to_drop, inplace=True)
```

```python
df
```

# Data Cleaning:
---
1. Deleting redundant columns.
2. Renaming the columns.
3. Dropping duplicates.
4. Cleaning individual columns.
5. Remove the NaN values from the dataset
6. Check for some more Transformations

```python
df.duplicated()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
modified_names = df.columns.str.capitalize()
```

```python
modified_names
```

```python
df.columns = modified_names
```

```python
df.dropna(inplace=True)
```

```python
df.reset_index(inplace=True)
```

```python
df.drop(columns='index', inplace=True)
```

```python
df['Rate'].tail()
```

```python
df['Rate'] = df['Rate'].str.replace(' ', '')
df['Rate'] = df['Rate'].str.replace('/', ' out of ')
```

```python
df['Online_order'] = df['Online_order'].apply(lambda x: True if x == 'Yes' else False)
df['Book_table'] = df['Book_table'] == 'Yes'
```

```python
df
```

```python
df.to_csv('Zomato_Dataset_Cleaned.csv', index=False)
```
